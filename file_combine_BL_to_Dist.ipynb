```python
# ==========================================================
# COMBINE BLOCK LEVEL WEATHER DATA TO DISTRICT LEVEL
# (Simple Average of Blocks)
# ==========================================================

from google.colab import files
import pandas as pd
import glob
import os

#------------------------------------------------------------
# 1. Upload all block files together
#------------------------------------------------------------
uploaded = files.upload()

#------------------------------------------------------------
# 2. Read all files
#------------------------------------------------------------
df_list = []

for file in uploaded.keys():

    print("Reading :", file)

    if file.lower().endswith(".csv"):
        df = pd.read_csv(file)

    elif file.lower().endswith((".xlsx",".xls")):
        df = pd.read_excel(file)

    else:
        continue

    df_list.append(df)

#------------------------------------------------------------
# 3. Combine all blocks
#------------------------------------------------------------
weather = pd.concat(df_list, ignore_index=True)

print("\nTotal Records :", len(weather))

#------------------------------------------------------------
# 4. Convert Date
#------------------------------------------------------------
weather["Date"] = pd.to_datetime(weather["Date"], errors="coerce")

#------------------------------------------------------------
# 5. Average block values to district level
#------------------------------------------------------------
district = (
    weather
    .groupby(["State","District","Date"], as_index=False)
    .agg({
        "Rainfall (mm)"            :"mean",
        "Max_Temperature (C)"      :"mean",
        "Min_Temperature (C)"      :"mean",
        "Relative_Humidity (%)"    :"mean",
        "Wind_Speed (m/s)"         :"mean",
        "Sunshine_Hours (hrs)"     :"mean",
        "Solar Radiation (mm/day)" :"mean"
    })
)

#------------------------------------------------------------
# 6. Round values
#------------------------------------------------------------
district = district.round(2)

#------------------------------------------------------------
# 7. Sort
#------------------------------------------------------------
district = district.sort_values(["State","District","Date"])

#------------------------------------------------------------
# 8. Save CSV
#------------------------------------------------------------
outfile = "District_Weather_Daily.csv"

district.to_csv(outfile,index=False)

print("\nFinished")
print("District Records :", len(district))
print("Districts :", district["District"].nunique())

#------------------------------------------------------------
# 9. Download
#------------------------------------------------------------
files.download(outfile)
```
