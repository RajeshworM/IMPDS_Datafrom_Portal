```python
#==========================================================
# DISTRICT WEATHER RANGE PERCENTAGE TABLE
# Google Colab (Single Run)
#==========================================================

import pandas as pd
import numpy as np
from google.colab import files

#---------------- Upload ----------------#
print("Upload Daily Weather CSV")
uploaded = files.upload()
file = list(uploaded.keys())[0]

#---------------- User Settings ----------------#

district = "District"
date = "Date"

# Bajra (Haryana) Crop Stages (ISO Week)

stages = {
    "Sowing": (24,29),
    "Vegetative": (30,34),
    "Reproductive": (35,38),
    "Maturity": (39,40),
    "Harvest": (41,42)
}

# Weather Ranges

variables = {

"Rainfall (mm)":("Rainfall",
[0,10,20,40,60,np.inf],
["0-10","10-20","20-40","40-60",">60"]),

"Max_Temperature (C)":("MaxTemp",
[-100,30,35,40,45,np.inf],
["<30","30-35","35-40","40-45",">45"]),

"Min_Temperature (C)":("MinTemp",
[-100,20,25,30,35,np.inf],
["<20","20-25","25-30","30-35",">35"]),

"Relative_Humidity (%)":("Humidity",
[0,40,60,80,90,101],
["<40","40-60","60-80","80-90",">90"]),

"Wind_Speed (m/s)":("Wind",
[0,1,2,4,6,np.inf],
["<1","1-2","2-4","4-6",">6"]),

"Sunshine_Hours (hrs)":("Sunshine",
[0,2,4,6,8,np.inf],
["<2","2-4","4-6","6-8",">8"]),

"Solar Radiation (mm/day)":("Solar",
[0,10,15,20,25,np.inf],
["<10","10-15","15-20","20-25",">25"])

}

#---------------- Read Data ----------------#

df = pd.read_csv(file)

df.columns = df.columns.str.strip()

print(df.columns.tolist())      # Check column names

df[date] = pd.to_datetime(df[date], dayfirst=True, errors="coerce")

df = df.dropna(subset=[date])

df["Week"] = df[date].dt.isocalendar().week.astype(int)

#---------------- Excel ----------------#

writer = pd.ExcelWriter("Weather_Percentage_Table.xlsx",
                        engine="openpyxl")

#---------------- Processing ----------------#

for var,(sheet,bins,labels) in variables.items():

    final=[]

    for stage,(w1,w2) in stages.items():

        # Works for both Kharif and Rabi
        if w1<=w2:
            temp=df[(df.Week>=w1)&(df.Week<=w2)].copy()
        else:
            temp=df[(df.Week>=w1)|(df.Week<=w2)].copy()

        temp=temp.dropna(subset=[var])

        temp["Range"]=pd.cut(temp[var],
                             bins=bins,
                             labels=labels,
                             include_lowest=True,
                             right=False)

        tab=(temp.groupby([district,"Range"],observed=True)
                  .size()
                  .reset_index(name="Days"))

        tab["Percent"]=100*tab["Days"]/tab.groupby(district)["Days"].transform("sum")

        wide=(tab.pivot(index=district,
                        columns="Range",
                        values="Percent")
                .round(1)
                .reset_index())

        wide.insert(1,"Stage",stage)

        final.append(wide)

    pd.concat(final,ignore_index=True).to_excel(writer,
                                                sheet_name=sheet,
                                                index=False)

writer.close()

print("Finished Successfully")

files.download("Weather_Percentage_Table.xlsx")
```

***For Rabi Crops**

```python
#=========================================================
# HARYANA WHEAT WEATHER RANGE PERCENTAGE TABLE
# Single Run Google Colab Program
#=========================================================

import pandas as pd
import numpy as np
from google.colab import files

#---------------------------------------------------------
# Upload CSV
#---------------------------------------------------------
print("Upload District Daily Weather CSV")
uploaded = files.upload()
file = list(uploaded.keys())[0]

#---------------------------------------------------------
# USER SETTINGS
#---------------------------------------------------------

district = "District"
date = "Date"

# Wheat crop stages (ISO Week)

stages = {

"Sowing"       :(42,44),

"Vegetative"   :(45,2),

"Reproductive" :(3,10),

"Maturity"     :(11,13),

"Harvest"      :(14,15)

}

#---------------------------------------------------------
# Wheat Weather Ranges
#---------------------------------------------------------

variables={

"Rainfall (mm)":
("Rainfall",
[0,2,5,10,20,np.inf],
["0-2","2-5","5-10","10-20",">20"]),

"Max_Temperature (C)":
("MaxTemp",
[-100,20,25,30,35,np.inf],
["<20","20-25","25-30","30-35",">35"]),

"Min_Temperature (C)":
("MinTemp",
[-100,5,10,15,20,np.inf],
["<5","5-10","10-15","15-20",">20"]),

"Relative_Humidity (%)":
("Humidity",
[0,40,60,80,90,101],
["<40","40-60","60-80","80-90",">90"]),

"Wind_Speed (m/s)":
("Wind",
[0,1,2,4,6,np.inf],
["<1","1-2","2-4","4-6",">6"]),

"Sunshine_Hours (hrs)":
("Sunshine",
[0,2,4,6,8,np.inf],
["<2","2-4","4-6","6-8",">8"]),

"Solar Radiation (mm/day)":
("Solar",
[0,8,12,16,20,np.inf],
["<8","8-12","12-16","16-20",">20"])

}

#---------------------------------------------------------
# Read Data
#---------------------------------------------------------

df=pd.read_csv(file)

df.columns=df.columns.str.strip()

df[date]=pd.to_datetime(df[date],dayfirst=True,errors="coerce")

df=df.dropna(subset=[date])

df["Week"]=df[date].dt.isocalendar().week.astype(int)

#---------------------------------------------------------
# Excel Writer
#---------------------------------------------------------

writer=pd.ExcelWriter("Haryana_Wheat_Weather_Percentage.xlsx",
                      engine="openpyxl")

#---------------------------------------------------------
# Process
#---------------------------------------------------------

for var,(sheet,bins,labels) in variables.items():

    output=[]

    for stage,(w1,w2) in stages.items():

        # Stage crossing year

        if w1<=w2:
            temp=df[(df["Week"]>=w1)&(df["Week"]<=w2)].copy()
        else:
            temp=df[(df["Week"]>=w1)|(df["Week"]<=w2)].copy()

        temp=temp.dropna(subset=[var])

        temp["Range"]=pd.cut(temp[var],
                             bins=bins,
                             labels=labels,
                             include_lowest=True,
                             right=False)

        tab=(temp
             .groupby([district,"Range"],observed=True)
             .size()
             .reset_index(name="Days"))

        tab["Percent"]=100*tab["Days"]/tab.groupby(district)["Days"].transform("sum")

        wide=(tab
              .pivot(index=district,
                     columns="Range",
                     values="Percent")
              .round(1)
              .reset_index())

        wide.insert(1,"Stage",stage)

        output.append(wide)

    final=pd.concat(output,ignore_index=True)

    final.to_excel(writer,
                   sheet_name=sheet,
                   index=False)

writer.close()

print("Finished Successfully")

files.download("Haryana_Wheat_Weather_Percentage.xlsx")
```
